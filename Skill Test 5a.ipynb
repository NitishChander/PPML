```python
#5A
import pandas as pd
import matplotlib.pyplot as plt

# Create a dummy sales_data.csv file if it doesn't exist
try:
    with open("sales_data.csv", "x") as f:
        f.write("Month,Sales\nJan,100\nFeb,120\nMar,150\nApr,130\nMay,180\nJun,200")
except FileExistsError:
    pass # File already exists, proceed

# 1. Read the CSV file
df = pd.read_csv("sales_data.csv")

# 2. Display first 5 rows
print("First 5 Rows of Dataset:")
print(df.head())

# 3. Check for missing values
print("\nMissing Values in Dataset:")
print(df.isnull().sum())

# Handle missing values by filling with 0
df.fillna(0, inplace=True)

# 4. Visualize monthly sales trend
plt.plot(df['Month'], df['Sales'], marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
